```python
# 19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).
array_pri = np.random.randint(1, 101, 20)
array_sec = np.random.randint(1, 101, 20)
produto_escalar = np.dot(array_pri,array_sec )

vetor1 = np.linalg.norm(array_pri)
vetor2 = np.linalg.norm(array_sec)
cosseno = produto_escalar / (vetor1 * vetor2)

angulo = np.arccos(cosseno)
print(f'array_pri {array_pri}')
print(f'array_sec {array_sec}')
print(f'Cosseno: {cosseno}')
print(f'Angulo: {angulo}')
print(f'Ângulo em graus: {np.degrees(angulo)}')
```

# 2. SciPy


```python
# 1. Calcule a integral da função f(x) = x² * sin(x)
#    no intervalo de 0 a π.

from scipy import integrate
def f(x):
  return x**2 * np.sin(x)

resultado, erro = integrate.quad(f, 0, np.pi)
print('Resultado da integral: ', resultado)
print('Erro estimado: ', erro)
```

```python
# 2. Resolva a equação diferencial:
#    dy/dx = x - y
#    com y(0) = 1, no intervalo de 0 a 10.

from scipy.integrate import solve_ivp
def eqDif(x, y):
    return x - y

sol = solve_ivp(eqDif, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

print(f'Tempos: {sol.t}')
print(f'Solução y: {sol.y[0]}')
```

```python
# 3. Encontre as raízes da função:
#    f(x) = x⁴ - 3x³ + 2
#    utilizando SciPy.

from scipy.optimize import root

def f(x):
  return x**4 - 3*x**3 + 2

# Verificação dos valores da função para identificar possíveis trocas de sinal.
for x in [-2, -1, 0, 1, 2, 3, 4]:
    print(f'x = {x}: f(x) = {f(x)}')

r1 = root(f, 1)
r2 = root(f, 2.5)

print(f'Raiz 1: {r1.x}')
print(f'Raiz 2: {r2.x}')
```

```python
# 4. Use a função de otimização do SciPy para maximizar:
#    f(x) = -x² + 4x + 1.
from scipy.optimize import minimize


def f_maximizar(x):
    return -(-x**2 + 4*x + 1)
resultado = minimize(f_maximizar, 0)
print(f'Máximo em x = {resultado.x}')
print(f'Valor máximo = {-resultado.fun}')
```

```python
# 5. Calcule a Transformada de Fourier de um sinal composto por:
#    sin(2πt) + cos(4πt).
from scipy.fft import fft
t = np . linspace (0, 1, 500)
sinal = np.sin (2 * np.pi * t) + np.cos (4 * np.pi * t)
frequencias = fft ( sinal )
print(np.abs(frequencias[:10]))
```

```python
# 6. Encontre a matriz inversa de uma matriz 5x5
#    de números aleatórios.
from scipy.linalg import inv


matriz_5x5 = np.random.rand(5,5)
inversa = inv(matriz_5x5)

print(f'Matriz:\n{matriz_5x5}')
print(f'Inversa:\n{inversa}')
print(f'Verificação (matriz × inversa):\n{np.dot(matriz_5x5, inversa)}')
```

```python
# 7. Use a interpolação cúbica do SciPy para interpolar
#    os pontos:
#    (0,0), (1,2), (2,3), (3,5), (4,4).
from scipy import interpolate
 # Definir os pontos
x = np.array ([0 , 1, 2 , 3 , 4])
y = np.array ([0 , 2, 3 , 5 , 4])

# Criar a fun o de interpolacao
f_interpolada = interpolate . interp1d (x, y, kind='cubic')

x_novos = np.array([0.5, 1.5, 2.5, 3.5])
y_novos = f_interpolada(x_novos)
print("Valores interpolados :", x_novos, y_novos )
```

```python
# 8. Resolva o sistema de equações:
#  3x + 2y = 5
#  x + 4y = 6

from scipy import linalg

A = np.array([[3, 2], [1, 4]])
B = np.array([5,6])

solucao = linalg.solve(A, B)

print(solucao)
```

```python
# 9. Ajuste uma curva quadrática aos pontos:
#    (1,2), (2,4), (3,5), (4,7), (5,8).
from scipy.optimize import curve_fit

def parabola(x, a, b, c):
    return a * x**2 + b * x + c

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

parametros, _ = curve_fit(parabola, x, y)

print(f'Parâmetros [a, b, c]: {parametros}')
```

```python
# 10. Calcule a integral dupla da função:
#     f(x, y) = xy + x²
#     com:
#     x ∈ [0, 2]
#     y ∈ [0, 1].

from scipy.integrate import dblquad

def f(y, x):
    return x*y + x**2

resultado, erro = dblquad(f, 0, 2, 0, 1)
print(resultado)
```

```python
# 14. Utilize a otimização multivariada para minimizar:
#     f(x, y) = (x - 2)² + (y - 3)².

def f(v):
    x, y = v
    return (x - 2)**2 + (y - 3)**2

minimize(f, [0, 0])

resultado = minimize(f, [0, 0])
print(f'Mínimo em [x, y] = {resultado.x}')
```

```python
# 15. Calcule a Transformada de Fourier Inversa
#     de um sinal complexo gerado aleatoriamente.

from scipy.fft import ifft
sinal_complexo = np.random.rand(50) + 1j * np.random.rand(50)

sinal_tempo = ifft(sinal_complexo)
print(np.allclose(fft(sinal_tempo), sinal_complexo))
print(f'Sinal complexo (5 primeiros):\n{sinal_complexo[:5]}')
print(f'Após ifft (5 primeiros):\n{sinal_tempo[:5]}')
```

```python
# 16. Resolva a equação diferencial de segunda ordem:
#     d²y/dx² + y = 0
#     com:
#     y(0) = 1
#     y'(0) = 0.

def eq(x, y):
    return [y[1], -y[0]]
sol = solve_ivp(eq, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 100), rtol=1e-8, atol=1e-10)
print(f'Solução bate com cos(x): {np.allclose(sol.y[0], np.cos(sol.t))}')
print(f'Solução y (10 primeiros): {sol.y[0][:10]}')
```

```python
# 18. Interpole a função:
#     f(x) = x²
#     nos pontos [1, 2, 3, 4, 5]
#     utilizando spline cúbica.

from scipy.interpolate import interp1d

x = np.array([1, 2, 3, 4, 5])
y = x **2

f_spline = interp1d(x, y, kind='cubic')

x_novos = np.array([1.5, 2.5, 3.5, 4.5])
y_novos = f_spline(x_novos)

print(f'Pontos originais: {y}')
print(f'Valores interpolados: {y_novos}')
```
