```python
# 2. Gere uma matriz 4x4 com numeros inteiros de 1 a 16 e extraia a ultima linha.

matriz_4x4 = np.arange(1, 17).reshape(4, 4)
ultima_linha = matriz_4x4[-1]

print(matriz_4x4)
print()
print(ultima_linha)

```

```python
# 3. Crie um array de zeros com comprimento 15 e substitua os valores da 5ª a 10ª posicao por 1.

array_zerado = np.zeros(15)
print(f'Array criado e zerado: {array_zerado}')

array_zerado[4:10] = 1
print(f'Array com o 1: {array_zerado}')
```

```python
# 4. Gere um array de 100 numeros inteiros aleatorios e encontre a mediana e a moda dos valores.

arrayInteiros = np.random.randint(1, 80, 100)
print(arrayInteiros)
mediana = np.median(arrayInteiros)
print(f'Mediana: {mediana}')
moda = np.argmax(np.bincount(arrayInteiros))
print(f'Moda: {moda}')
```

```python
# 5. Crie um array de 20 elementos contendo numeros entre 1 e 100 e substitua todos os valores multiplos de 3 por -3.
arrayelementos = np.random.randint(1, 101, 20)
print(f'Array antes de mudar: {arrayelementos}')
arrayelementos[arrayelementos % 3 == 0] = -3
print()
print(f'Array apos mudar: {arrayelementos}')

```

```python
# 6. Gere uma matriz 5x5 de numeros aleatorios entre 0 e 1 e encontre a soma de cada coluna.
matriz_5x5 = np.random.rand(5,5)
print(f'Matriz:\n{matriz_5x5}')

somaColunas = matriz_5x5.sum(axis=0)
print(f'Soma das colunas: {somaColunas}')
```

```python
# 7. Crie dois arrays de numeros aleatorios e encontre a correlacao entre eles
arrayePrimario = np.random.randint(1, 30, 20)
arrayeSecundario = np.random.randint(1, 30, 20)
correlacao = np.corrcoef(arrayePrimario, arrayeSecundario) [0, 1]

print(f'Primeiro: {arrayePrimario}')
print(f'Secundario: {arrayeSecundario}')
print(f'Correlacao: {correlacao}')
```

```python
# 8. Gere uma matriz 3x3 com numeros inteiros aleatorios e eleve todos os seus elementos ao quadrado.

matriz_3x3 = np.random.randint(1, 10, (3, 3))
print(f'Matriz:\n{matriz_3x3}')
elevados = matriz_3x3 ** 2
print(f'Quadrados:\n{elevados}')
```

```python
# 9. Crie um array com 10 numeros aleatorios e normalize seus valores entre 0 e 1
array_10 = np.random.randint(1, 20, 10)
print(array_10)

normalizado = (array_10 - array_10.min()) / (array_10.max() - array_10.min())
print(normalizado)

print(normalizado.min())   # tem que dar 0.0
print(normalizado.max())   # tem que dar 1.0
```

```python
# 10. Crie uma matriz identidade 6x6 e substitua a diagonal principal por valores crescentes de 1 a 6.
matriz_6x6 = np.eye(6)
print(matriz_6x6)
np.fill_diagonal(matriz_6x6, [1,2,3,4,5,6])
print(matriz_6x6)
```

```python
# 11. Gere um array de 100 números inteiros e encontre o valor que mais se repete no array.
array_100 = np.random.randint(1, 100, 100)
print((f'array original:\n{array_100}'))
repete = np.argmax(np.bincount(array_100))
print(f'O numero que mais se repete: {repete}')

# bincount conta quantas vezes cada número aparece;
# o índice do resultado é o próprio número, então argmax devolve a moda
```

```python
# 12. Crie uma matriz 4x4 e substitua os valores da diagonal principal por 7.
matriz_4x4 = np.random.randint(1, 60, (3, 3))
print(f'Matriz original:\n{matriz_4x4}')
np.fill_diagonal(matriz_4x4, [7])
print(f'Matriz mudada:\n {matriz_4x4}')
```

```python
# 13. Crie um array contendo os 10 primeiros números pares e calcule sua soma cumulativa.
array_pares = np.arange(2, 21, 2)
print(f'Pares: {array_pares}')

somas = np.cumsum(array_pares)
print(somas)
```

```python
# 14. Calcule a diferença entre dois arrays de números aleatórios de mesmo comprimento e encontre a média das diferenças.
arrayePrimario = np.random.randint(1, 30, 20)
arrayeSecundario = np.random.randint(1, 30, 20)
print(f'Array1: {arrayePrimario}')
print(f'Array2: {arrayeSecundario}')
diferenca = arrayePrimario - arrayeSecundario
media = diferenca.mean()
print(f'Media das diferencas: {media}')
```

```python
# 17. Gere uma matriz 4x4 de números inteiros aleatórios e multiplique cada valor por sua posição na matriz.
matriz_4x4 = np.random.randint(1, 10, (4, 4))
print(f'Matriz:\n{matriz_4x4}')
matriz_zerada = np.arange(0,16).reshape(4, 4)
print(f'Matriz:\n{matriz_zerada}')

multMatriz = matriz_4x4 * matriz_zerada
print(f'Matriz multiplicada:\n{multMatriz}')
```

```python
# 18. Crie um array de 10 elementos contendo números entre 1 e 100 e substitua todos os valores múltiplos de 5 por 0.
array_10elementos = np.random.randint(1, 101, 10)
print(f'Array:\n{array_10elementos}')
array_10elementos[array_10elementos % 5 == 0] = 0
print(f'Array multiplos de 5 retirado:\n{array_10elementos}')
```

```python
# 19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).
array_pri = np.random.randint(1, 101, 20)
array_sec = np.random.randint(1, 101, 20)
produto_escalar = np.dot(array_pri,array_sec )

vetor1 = np.linalg.norm(array_pri)
vetor2 = np.linalg.norm(array_sec)
cosseno = produto_escalar / (vetor1 * vetor2)

angulo = np.arccos(cosseno)
print(f'array_pri {array_pri}')
print(f'array_sec {array_sec}')
print(f'Cosseno: {cosseno}')
print(f'Angulo: {angulo}')
print(f'Ângulo em graus: {np.degrees(angulo)}')
```
