```python
# 1. Calcule a integral da função f(x) = x² * sin(x)
#    no intervalo de 0 a π.
# usei o quad que ja faz a integral numerica sozinho, ele devolve o resultado e tambem uma estimativa de erro, entao imprimi os dois.

from scipy import integrate
def f(x):
  return x**2 * np.sin(x)

resultado, erro = integrate.quad(f, 0, np.pi)
print('Resultado da integral: ', resultado)
print('Erro estimado: ', erro)
```

```python
# 2. Resolva a equação diferencial:
#    dy/dx = x - y
#    com y(0) = 1, no intervalo de 0 a 10.

# usei solve_ivp pra resolver a equacao diferencial, o t_eval e so pra falar em quais pontos eu quero

from scipy.integrate import solve_ivp
def eqDif(x, y):
    return x - y

sol = solve_ivp(eqDif, [0, 10], [1], t_eval=np.linspace(0, 10, 100))

print(f'Tempos: {sol.t}')
print(f'Solução y: {sol.y[0]}')
```

```python
# 3. Encontre as raízes da função:
#    f(x) = x⁴ - 3x³ + 2
#    utilizando SciPy.

# antes de usar o root fui testando o sinal da funcao em varios x pra ver onde ela troca de sinal,
# porque isso indica que tem raiz ali perto, ai usei esses valores como chute inicial.

from scipy.optimize import root

def f(x):
  return x**4 - 3*x**3 + 2

# Verificação dos valores da função para identificar possíveis trocas de sinal.
for x in [-2, -1, 0, 1, 2, 3, 4]:
    print(f'x = {x}: f(x) = {f(x)}')

r1 = root(f, 1)
r2 = root(f, 2.5)

print(f'Raiz 1: {r1.x}')
print(f'Raiz 2: {r2.x}')
```

```python
# 4. Use a função de otimização do SciPy para maximizar:
#    f(x) = -x² + 4x + 1.

# o minimize so sabe minimizar, entao pra maximizar eu inverti o sinal da funcao (coloquei um menos na
# frente) e no final devolvi o valor com o sinal certo de novo.

from scipy.optimize import minimize


def f_maximizar(x):
    return -(-x**2 + 4*x + 1)
resultado = minimize(f_maximizar, 0)
print(f'Máximo em x = {resultado.x}')
print(f'Valor máximo = {-resultado.fun}')
```

```python
# 5. Calcule a Transformada de Fourier de um sinal composto por:
# sin(2πt) + cos(4πt).

# usei fft pra decompor o sinal em frequencias, e abs() pra pegar so o tamanho (magnitude) de cada
# uma, ignorando a parte complexa.
from scipy.fft import fft
t = np . linspace (0, 1, 500)
sinal = np.sin (2 * np.pi * t) + np.cos (4 * np.pi * t)
frequencias = fft ( sinal )
print(np.abs(frequencias[:10]))
```

```python
# 6. Encontre a matriz inversa de uma matriz 5x5 de números aleatórios.

# usei inv pra achar a inversa e depois multipliquei a matriz pela inversa de proposito, se der certo
# o resultado tem que ficar bem proximo da identidade.

from scipy.linalg import inv


matriz_5x5 = np.random.rand(5,5)
inversa = inv(matriz_5x5)

print(f'Matriz:\n{matriz_5x5}')
print(f'Inversa:\n{inversa}')
print(f'Verificação (matriz × inversa):\n{np.dot(matriz_5x5, inversa)}')
```

```python
# 7. Use a interpolação cúbica do SciPy para interpolar os pontos:
# (0,0), (1,2), (2,3), (3,5), (4,4).

# usei interp1d com kind='cubic' pra ele passar uma curva suave exatamente por cima dos pontos que eu tinha.

from scipy import interpolate

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])

f_interpolada = interpolate . interp1d (x, y, kind='cubic')

x_novos = np.array([0.5, 1.5, 2.5, 3.5])
y_novos = f_interpolada(x_novos)
print(f'x: {x_novos}')
print(f'y interpolado: {y_novos}')
```

```python
# 8. Resolva o sistema de equações:
#  3x + 2y = 5
#  x + 4y = 6

# usei linalg.solve porque resolve o sistema direto, sem precisar ficar calculando a inversa da matriz na mao.

from scipy import linalg

A = np.array([[3, 2], [1, 4]])
B = np.array([5,6])

solucao = linalg.solve(A, B)

print(solucao)
```

```python
# 9. Ajuste uma curva quadrática aos pontos:
#    (1,2), (2,4), (3,5), (4,7), (5,8).

# usei curve_fit passando a funcao da parabola, ele mesmo acha os parametros a, b e c que encaixam melhor nos pontos.
from scipy.optimize import curve_fit

def parabola(x, a, b, c):
    return a * x**2 + b * x + c

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

parametros, _ = curve_fit(parabola, x, y)

print(f'Parâmetros [a, b, c]: {parametros}')
```

```python
# 10. Calcule a integral dupla da função:
#     f(x, y) = xy + x²
#     com:
#     x ∈ [0, 2]
#     y ∈ [0, 1].

# usei dblquad pra integral dupla, so tive que prestar atencao que a funcao tem que receber (y, x)
# nessa ordem por causa de como a funcao pede.

from scipy.integrate import dblquad

def f(y, x):
    return x*y + x**2

resultado, erro = dblquad(f, 0, 2, 0, 1)
print(resultado)
```

```python
# 11. Encontre os autovalores e autovetores
#     de uma matriz 4x4 de números inteiros aleatórios.

# usei eig que ja devolve autovalor e autovetor junto, nao precisei calcular na mao.
from scipy import linalg
matriz_4x4 = np.random.randint(1, 40, (4, 4))

autovalores, autovetores = linalg.eig(matriz_4x4)

print(f'Autovalores:\n{autovalores}')
print(f'Autovetores:\n{autovetores}')
```

```python
# 12. Resolva a equação:
#     e^x - x² = 0
#     encontrando a raiz mais próxima de x = 1.

# usei fsolve dando um chute perto de 1, ele vai ajustando ate achar a raiz mais proxima desse chute.

from scipy.optimize import fsolve

def f(x):
    return np.exp(x) - x**2

raiz = fsolve(f, 1)
print(f'Raiz: {raiz}')
```

```python
# 14. Utilize a otimização multivariada para minimizar:
#     f(x, y) = (x - 2)² + (y - 3)².

# aqui o minimize recebe duas variaveis dentro de uma lista (x e y) e ele acha o ponto onde os dois
# juntos deixam a funcao minima.

def f(v):
    x, y = v
    return (x - 2)**2 + (y - 3)**2

resultado = minimize(f, [0, 0])
print(f'Mínimo em [x, y] = {resultado.x}')
```

```python
# 15. Calcule a Transformada de Fourier Inversa
#     de um sinal complexo gerado aleatoriamente.

# usei ifft pra voltar do dominio de frequencia pro sinal original, e conferi com allclose se bate com
# o sinal que eu tinha antes de aplicar a fft.

from scipy.fft import ifft
sinal_complexo = np.random.rand(50) + 1j * np.random.rand(50)

sinal_tempo = ifft(sinal_complexo)
print(np.allclose(fft(sinal_tempo), sinal_complexo))
print(f'Sinal complexo (5 primeiros):\n{sinal_complexo[:5]}')
print(f'Após ifft (5 primeiros):\n{sinal_tempo[:5]}')
```

```python
# 16. Resolva a equação diferencial de segunda ordem:
#     d²y/dx² + y = 0
#     com:
#     y(0) = 1
#     y'(0) = 0.

# como e uma equacao de segunda ordem, transformei em duas de primeira ordem (é o jeito que o
# solve_ivp aceita) e depois conferi comparando com cos(x), que e a resposta que eu ja esperava.

def eq(x, y):
    return [y[1], -y[0]]
sol = solve_ivp(eq, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 100), rtol=1e-8, atol=1e-10)
print(f'Solução bate com cos(x): {np.allclose(sol.y[0], np.cos(sol.t))}')
print(f'Solução y (10 primeiros): {sol.y[0][:10]}')
```

```python
# 6. Plote dois graficos de linha em subplots diferentes: y1 = e^x e y2 = log(x).
# usei subplot(2,1,1) e subplot(2,1,2) pra desenhar os dois graficos um embaixo do outro na mesma figura.


x = np.linspace(0.1, 5, 100)
y1 = np.exp(x)
y2 = np.log(x)

plt.subplot(2, 1, 1)     # 2 linhas, 1 coluna, 1º gráfico
plt.plot(x, y1)
plt.title('Função Exponencial (e^x)')

plt.subplot(2, 1, 2)     # 2º gráfico
plt.plot(x, y2)
plt.title('Função Logarítmica (log(x))')

plt.tight_layout()
plt.show()
```

```python
# 8. Plote um gráfico de linha com intervalos de confiança (sombreamento) para a função $y = \sin(x)$.
# usei fill_between pra pintar a area entre y-erro e y+erro, dando a ideia visual de uma margem em volta da linha.

x = np.linspace(0, 10, 100)
y = np.sin(x)
erro = 0.2 # Um valor escolhido, já que o enunciado não define a margem

plt.plot(x, y)
plt.fill_between(x, y - erro, y + erro, alpha=0.3)
plt.title("Função Seno com Sombreamento")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
