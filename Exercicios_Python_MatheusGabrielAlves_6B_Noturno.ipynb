```python
# 21 - Implemente a busca binária para encontrar um elemento em uma lista ordenada.

numeros = [3, 7, 12, 18, 24, 31, 45, 52, 60, 67, 73, 81, 88, 94, 99]

procurado = int(input("Digite o número que deseja procurar: "))

inicio = 0
fim = len(numeros) - 1 # Define o fim do intervalo usando a última posição da lista

while inicio <= fim: # Utilizei o while para continuar a busca enquanto ainda tiver posições para verificar.
    meio = (inicio + fim) // 2 # Utilizei para encontrar a posição do meio do intervalo da busca.

    if numeros[meio] == procurado:  # Verifica se o número que está no meio é o número procurado
        print("Número encontrado na posição:", meio)
        break  # Utilizei para encerrar o while assim que encontrar o número.

    elif procurado < numeros[meio]: # Verifica se o número procurado está antes do número que está no meio.
        fim = meio - 1 # Diminui o intervalo para continuar a busca somente na parte esquerda.

    else:
        inicio = meio + 1 # Caso o número seja maior que o do meio, continua a busca na parte direita

if inicio > fim:
    print("Número não encontrado.")
```

```python
# 22 - Crie uma função para criptografar uma mensagem usando a cifra de César.

def criptografar(mensagem, deslocamento):
    alfabeto = "abcdefghijklmnopqrstuvwxyz" #Criei o alfabeto para conseguir encontrar a posição de cada letra
    resultado = ""

    for letra in mensagem: # Usei o for para percorrer a mensagem.
        if letra in alfabeto:
            posicao = alfabeto.index(letra)  # Encontra a posição da letra dentro do alfabeto.
            nova_posicao = (posicao + deslocamento) % 26 # O deslocamento define quantas posições a letra vai avançar no alfabeto e o % 26 faz voltar ao início quando passar da letra z
            resultado += alfabeto[nova_posicao] # Adiciona ao resultado a letra que foi encontrada na nova posição.
        else:
            resultado += letra # Mantém caracteres que não estão no alfabeto, como espaços e pontuação.


    return resultado


mensagem = input("Digite uma mensagem: ").lower()
deslocamento = int(input("Digite o deslocamento: "))

print(criptografar(mensagem, deslocamento))
```

```python
# 23 - Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.

def descriptografar(mensagem, deslocamento):
    alfabeto = "abcdefghijklmnopqrstuvwxyz" # Criei uma string com o alfabeto para usar como referência e achar a posição de cada letra
    resultado = ""

    for letra in mensagem: # percorro a mensagem com o for e subtraio o deslocamento da posição para voltar à letra original,
        if letra in alfabeto:
            posicao = alfabeto.index(letra)
            nova_posicao = (posicao - deslocamento) % 26 % 26  # Subtraio o deslocamento para voltar à posição da letra original.
            resultado += alfabeto[nova_posicao]
        else:
            resultado += letra

    return resultado


mensagem = input("Digite a mensagem criptografada: ").lower()
deslocamento = int(input("Digite o deslocamento: "))

print(descriptografar(mensagem, deslocamento))
```

```python
# 24 - Implemente uma função para validar um CPF brasileiro.

def validar_cpf(cpf):
    cpf = cpf.replace(".", "").replace("-", "")  # Removi pontos e traços para trabalhar somente com os números.

    # Organizei as validações do mais simples para o cálculo, evitando erros antes de usar int().

    if len(cpf) != 11:
        return False

    if not cpf.isdigit():  # Verifico se o CPF contém somente números.
        return False

    if cpf == cpf[0] * 11:  # Rejeito CPFs formados pelo mesmo número repetido.
        return False

    # Aqui começam os cálculos dos dígitos verificadores.
    soma = 0

    for i in range(9):  # Usei o for para fazer o cálculo do primeiro dígito verificador.
        soma += int(cpf[i]) * (10 - i)

    resto = soma % 11

    if resto < 2:
        primeiro = 0
    else:
        primeiro = 11 - resto

    if primeiro != int(cpf[9]):  # Verifico se o primeiro dígito calculado é igual ao informado no CPF.
        return False

    soma = 0

    for i in range(10):  # Repito o cálculo para encontrar o segundo dígito verificador.
        soma += int(cpf[i]) * (11 - i)

    resto = soma % 11

    if resto < 2:
        segundo = 0
    else:
        segundo = 11 - resto

    if segundo != int(cpf[10]):  # Verifico se o segundo dígito calculado é igual ao informado no CPF.
        return False

    return True  # Retorno True quando os dois dígitos verificadores estão corretos.


cpf = input("Digite o CPF: ")

if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

```python
# 27 -  Verifique se duas palavras são anagramas (ou seja, se são formadas pelas mesmas letras).

#Fiz uma função para verificar, que de início já deixa as palavras em minúsculo para não ter problemas.
# Usei o len para calcular a quantidade de letras de cada palavra e já eliminar os casos de tamanhos diferentes.
# Se for a mesma quantidade, utilizo o sorted para organizar as letras e com isso saber se são anagramas.
def verificarAnagramas(palavra1, palavra2):
  palavra1 = palavra1.lower()
  palavra2 = palavra2.lower()
  qtd1 =len(palavra1)
  qtd2 = len(palavra2)
  if qtd1 != qtd2:
    return False
  if sorted(palavra1) == sorted(palavra2):
    return True

  return False
palavra1 = input('Digite uma palavra: ')
palavra2 = input('Digite outra palavra: ')

if verificarAnagramas(palavra1, palavra2):
    print('São anagramas')
else:
    print('Não são anagramas')
```

```python
# 30 - Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra.
#Utilizei uma função para criar o jogo da forca.
def jogarForca():
    palavra = "python"
    letrasAcertadas = [] # Usei uma lista para armazenar as letras acertadas
    tentativas = 6

    while tentativas > 0: # While para repetir o jogo enquanto ainda houver tentativas.
        acertou = True

        for letra in palavra: # O for percorre a palavra mostrando as letras já descobertas
            if letra in letrasAcertadas:
                print(letra, end=" ")
            else:
                print("_", end=" ") # _ para as que ainda não foram acertadas
                acertou = False

        print()

        if acertou:
            print("Parabéns, você venceu!")
            return

        tentativa = input("Digite uma letra: ").lower()

        if tentativa in palavra: # Quando o usuário acerta uma letra, ela é adicionada na lista, caso contrário, o número de tentativas é reduzido.
            letrasAcertadas.append(tentativa)
            print("Você acertou uma letra!")
        else:
            tentativas -= 1
            print("Letra incorreta. Restam", tentativas, "tentativas.")

    print("Você perdeu!")
    print("A palavra era:", palavra)
#No final, o programa informa se o usuário venceu ou perdeu

jogarForca()
```

```python
# 31 - Implemente o algoritmo de ordenação por bolha (Bubble Sort)
# Utilizei uma função para ordenar a lista usando o Bubble Sort.
def bubbleSort(lista):
    tamanho = len(lista)

    for i in range(tamanho): # Usei dois laços for, onde o primeiro controla a quantidade de passadas e o segundo compara os elementos vizinhos da lista.
        for j in range(0, tamanho - i - 1):
            if lista[j] > lista[j + 1]: # Quando um elemento é maior que o próximo, eles trocam de posição.
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista

# Ao final das comparações, a lista fica ordenada em ordem crescente.
numeros = [8, 3, 5, 1, 9, 2]

print("Lista original:", numeros)
print("Lista ordenada:", bubbleSort(numeros))
```

```python
# 32 - Crie um jogo da velha para dois jogadores.

def mostrarTabuleiro(tabuleiro): # Utilizei o for para percorrer cada linha da matriz e mostrar o tabuleiro.
    for linha in tabuleiro:
        print(" | ".join(linha))
        print("-" * 9)


def verificarVencedor(tabuleiro, jogador): # Utilizei o for para verificar as linhas e colunas do tabuleiro.
    for i in range(3):
        if tabuleiro[i][0] == jogador and tabuleiro[i][1] == jogador and tabuleiro[i][2] == jogador:
            return True

        if tabuleiro[0][i] == jogador and tabuleiro[1][i] == jogador and tabuleiro[2][i] == jogador:
            return True

    if tabuleiro[0][0] == jogador and tabuleiro[1][1] == jogador and tabuleiro[2][2] == jogador: # Verifica se o jogador completou a diagonal principal.
        return True

    if tabuleiro[0][2] == jogador and tabuleiro[1][1] == jogador and tabuleiro[2][0] == jogador: # Verifica se o jogador completou a diagonal secundária.
        return True

    return False


tabuleiro = [
    [" ", " ", " "],
    [" ", " ", " "],
    [" ", " ", " "]
]

jogador = "X"

for rodada in range(9): # Utilizei o for para controlar as 9 possíveis jogadas do jogo.
    mostrarTabuleiro(tabuleiro)

    linha = int(input("Digite a linha (0 a 2): "))
    coluna = int(input("Digite a coluna (0 a 2): "))

    if tabuleiro[linha][coluna] == " ": # Verifica se a posição escolhida está livre.
        tabuleiro[linha][coluna] = jogador

        if verificarVencedor(tabuleiro, jogador): # Verifica se o jogador venceu depois de fazer a jogada.
            mostrarTabuleiro(tabuleiro)
            print("Jogador", jogador, "venceu!")
            break

        if jogador == "X":
            jogador = "O"
        else:
            jogador = "X" # Verifica se o jogador venceu depois de fazer a jogada.

    else:
        print("Posicao já ocupada.")

else:
    mostrarTabuleiro(tabuleiro)
    print("Empate")
```

```python
# 33 - Verifique se duas palavras fornecidas são anagramas uma da outra.
# Escolhi utilizar uma função para organizar o código e deixar a verificação das palavras separada do restante do programa.
def verificador(palavra1, palavra2):
  palavra1 = palavra1.lower()
  palavra2 = palavra2.lower()
  qtd1 =len(palavra1)
  qtd2 = len(palavra2)
  if qtd1 != qtd2:  # Verifica se as palavras possuem a mesma quantidade de letras.
    return False
  if sorted(palavra1) == sorted(palavra2):  # Ordena as letras e verifica se são iguais nas duas palavras.
    return True

  return False # Retorna falso caso as palavras não sejam anagramas.
palavra1 = input('Digite uma palavra: ')
palavra2 = input('Digite outra palavra: ')

if verificador(palavra1, palavra2):
    print('São anagramas')
else:
    print('Não são anagramas')
```

```python
# 36 - Crie um gerador de senhas aleatórias com letras e números.
# Utilizei a biblioteca random para gerar caracteres aleatórios.
import random

#Criei uma função para montar a senha e uma variável contendo letras e números.
def gerarSenha(tamanho):
    caracteres = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
    senha = ""

    for i in range(tamanho): # Usei o for para repetir a quantidade de caracteres informada pelo usuário e o random.choice() para escolher um caractere aleatório a cada repetição.
        senha += random.choice(caracteres)

    return senha


tamanho = int(input("Digite o tamanho da senha: "))

print("Senha gerada:", gerarSenha(tamanho))
```

```python
# 37 - Implemente uma função recursiva para calcular o fatorial de um número.

def calcularFatorial(numA):
    if numA == 0 or numA == 1:  # Quando chegar em 0 ou 1, a recursão para.
        return 1

    return numA * calcularFatorial(numA - 1)  # A função chama ela mesma usando o número anterior.

try:
    numA = int(input("Digite um número: "))
    fatorial = calcularFatorial(numA)

    print("O fatorial é:", fatorial)

except ValueError:
    print("Digite apenas valores numéricos.")
```

```python
# 48 - Implemente o algoritmo de ordenação por inserção (Insertion Sort).

lista = [8, 3, 5, 1, 9, 2]

for i in range(1, len(lista)): # Começo pelo segundo elemento, pois considero o primeiro já ordenado.
    atual = lista[i] # Armazena o valor atual que será colocado na posição correta.
    j = i - 1

    while j >= 0 and lista[j] > atual: # Enquanto houver elementos anteriores maiores que o atual, continuo deslocando eles.
        lista[j + 1] = lista[j]
        j -= 1

    lista[j + 1] = atual # Coloca o valor atual na posição correta.

print(lista)
```
