## Exercício 03 — Área do Círculo

**Abordagem:**

- **math.pi:** utilizei para representar o valor de π com maior precisão, tornando o cálculo da área mais preciso e o código mais legível.
- **try/except:** utilizei para tratar possíveis erros de entrada. Caso o usuário digite um valor inválido, o programa não será interrompido e exibirá uma mensagem informando o erro.
- **float:** utilizei porque o raio pode possuir valores decimais.
- **round:** utilizei para exibir o resultado da área com apenas duas casas decimais, melhorando esteticamente.

```python
#3 - Solicite o raio de um círculo e calcule sua área. Use a fórmula área =π * r².

import math
try:
  raio = float(input('Digite o raio: '))
  area = math.pi * raio**2
  print('Valor da área:', round(area, 2))
except ValueError:
  print('Digite apenas valores numericos.')
```

```python
#4. Conversão de Temperatura  o Converta uma temperatura fornecida em Celsius para Fahrenheit  usando a fórmula F = C * 9/5 + 32.
#Mesmo raciocinio do exercicio 3, so mudei o round para apenas uma casa decimal ja que se trata de graus
try:
  grauC = float(input('Digite os graus celsius '))
  grauF = grauC * 9/5 + 32
  print('O valor em Fahrenheit é: ',round(grauF, 1))
except ValueError:
  print('Digite apenas valores numericos.')
```

*Abordagem:**

- **`for`:** decidi utilizar o `for` devido saber quantas vezes quero repetir. Como é uma tabuada, então o mais convencional foi até o 10.
- **`range(1, 11)`:** foi que se caso utilizasse `range(1, 10)` iria fazer só até o 9 e não repetiria toda a tabuada.
- **Demais decisões:** `try/except` e conversão de entrada seguem o mesmo padrão adotado no exercício 3.

```python
#Crie um programa que imprima a tabuada de um número fornecido pelo usuário.

try:
    numA = int(input('Digite um numero, e veja a sua tabuada: '))
    for contador in range(1, 11):
      resultado = (numA * contador)
      print(numA, "x", contador, "=", resultado)
except ValueError:
    print('Digite apenas valores numericos.')
```
